# 10. Swap two numbers (with third variable)

```python
num1 = int(input("Enter the First number :"))
num2 = int(input("Enter the Second number :"))

print("Before swapping:")
print("number 1", num1)
print("number 2",num2)

temp = num1
num1 = num2
num2 = temp

print("After Swapping :")
print("number 1", num1)
print("number 2",num2)
```
