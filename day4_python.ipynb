# 2.Count vowels

```python
word = input("enter a word: ")
count = 0
vowels = ["a","e","i","o","u","A","E","I","O","U"]

for letter in word:
    if letter in vowels:
        count = count + 1

print("Number of vowels:", count)
```

# 6.Find frequency of characters

```python
freq = str(input("enter a word"))
count = {}
for char in freq:
    if char in count:
        count[char] += 1
    else:
        count[char] = 1
print(count)
```
