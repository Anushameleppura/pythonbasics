# 6.Count digits in a number

```python
digit = int(input("Enter the digit :"))

count = 0
while digit != 0:
    digit = digit // 10
    count = count + 1
    print(count)
```

# 7.Reverse a number

```python
num = int(input("enter a number"))

reverse = 0
while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10
print(reverse)
```

# 8.Sum of digits

```python
num = int(input("Enter a number :"))

total = 0
while num != 0:
    digit = num % 10
    total = total + digit
    num = num // 10
print(total)
```

# 9.Fibonacci series

```python
num = int(input("Enter a Number :"))

first = 0
second =1

print(first)
print(second)

for i in range(num - 2):
    next_number = first + second
    print(next_number)
    
    first = second
    second = next_number
```

# 10.Check prime number

```python
num = int(input("Enter a number: "))

if num <= 1:
    print("Not prime")
else:
    is_prime = True

    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        print("Prime")
    else:
        print("Not prime")
```
